D: Amali matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_d = {
    "Nama": ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"],
    "Markah": [70, 85, 90, 65, 80],
    "Gred": ["B", "A", "A", "C", "B"]
}
df = pd.DataFrame(data_d)
```

```python
# Langkah 1: Bina Graf Bar Markah Pelajar
```

```python
plt.figure(figsize=(6, 4))
plt.bar(df["Nama"], df["Markah"])
plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")
plt.show()
```

Bina Graf Garisan

```python
plt.figure(figsize=(6, 4))
plt.plot(df["Nama"], df["Markah"], marker="o")
plt.title("Prestasi Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")
plt.show()
```

Bina Histogram

```python
plt.figure(figsize=(6, 4))
plt.hist(df["Markah"], bins=5)
plt.title("Taburan Markah Pelajar")
plt.xlabel("Julat Markah")
plt.ylabel("Bilangan Pelajar")
plt.show()
```



Bina Carta Pai Berdasarkan Gred



```python
jumlah_gred = df["Gred"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    jumlah_gred,
    labels=jumlah_gred.index,
    autopct="%1.1f%%"
)
plt.title("Peratusan Gred Pelajar")
plt.show()
```

Bahagian E (Gabungan NumPy, Pandas, dan Matplotlib)
Python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Nama": ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"],
    "Markah": [70, 85, 90, 65, 80]
}
df = pd.DataFrame(data)
```

Lakukan pengiraan menggunakan NumPy

```python
purata = np.mean(df["Markah"])
tertinggi = np.max(df["Markah"])
terendah = np.min(df["Markah"])

print("--- Hasil Pengiraan NumPy ---")
print("Purata:", purata)
print("Markah tertinggi:", tertinggi)
print("Markah terendah:", terendah)
print("-" * 30)
```

Tambah status menggunakan pandas

```python
df["Status"] = np.where(df["Markah"] >= 50, "Lulus", "Gagal")
print("\n--- Jadual Data Pelajar (DataFrame) ---")
print(df)
print("-" * 30)
```

Paparkan data menggunakan matplotlib

```python
plt.figure(figsize=(7, 4))
plt.bar(df["Nama"], df["Markah"], color="skyblue")
plt.axhline(
    y=purata,
    color="red",
    linestyle="--",
    label=f"Purata ({purata:.1f})"
)
plt.title("Analisis Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")
plt.legend()
plt.show()
```
