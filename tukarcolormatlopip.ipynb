```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

plt.bar(nama, markah, color="blue")
plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")
plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

plt.bar(
    nama,
    markah,
    color="skyblue",
    edgecolor="black",
    linewidth=2
)

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

# Menetapkan warna latar belakang kawasan graf
plt.gca().set_facecolor("lightyellow")

plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

warna = []


for nilai in markah:
    if nilai >= 80:
        warna.append("green")
    else:
        warna.append("red")

plt.bar(nama, markah, color=warna)
plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")
plt.show()
```
