```python

import pandas as pd

data = {
    "Nama": ["ali","ameer","alim","kasim"],
    "Markah" : [70,85,90,65]
}


```

BINA DATAFRAME

```python
df = pd.DataFrame(data)
print(df)
```

purata = df["Markah"].mean()
print("Purata markah:",purata)

```python
purata = df["Markah"].mean()
print("Purata markah:",purata)
```

tambah lajur status

```python
pelajar_cemerlang = df[df["Markah"] < 50]
print("Pelajar yang cemerlang:")
print(pelajar_cemerlang)
```

```python
df["status"] = np.where(df["Markah"] < 50, "cemerlang", "lulus")
print(df)
```

```python
def tentukan_gred(markah):
    if markah >= 80:
        return "A"
    elif markah >= 70:
        return "B"
    elif markah >= 60:
        return "C"
    elif markah >= 50:
        return "D"
    else:
        return "E"


df["Grade"] = df["Markah"].apply(tentukan_gred)
print(df)
```

```python
df.to_csv("data_pelajar.csv", index=False)

print("markah_pelajar.csv" , index=False)
```
