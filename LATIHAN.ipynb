Meminta Nama dan Umur Pengguna

```python

nama = input("Masukkan nama anda: ")
umur = input("Masukkan umur anda: ")


print(f"Nama: {nama}")
print(f"Umur: {umur} tahun")
```

Mengira Luas Segi Empat

```python
panjang = float(input("Masukkan panjang segi empat: "))
lebar = float(input("Masukkan lebar segi empat: "))

luas = panjang * lebar

print(f"Luas segi empat adalah: {luas}")
```

Menentukan Nombor Ganjil atau Genap

```python
nombor = int(input("Masukkan satu nombor: "))

if nombor % 2 == 0:
    print(f"Nombor {nombor} ialah nombor GENAP.")
else:
    print(f"Nombor {nombor} ialah nombor GANJIL.")
```

Memaparkan Nombor 1 hingga 10 Menggunakan Gelung (Loop)

```python
for i in range(1, 11):
    print(i)
```

Menentukan Status Lulus atau Gagal Berdasarkan Markah

```python
markah = float(input("Masukkan markah anda: "))

if markah >= 40:
    print("Status: LULUS")
else:
    print("Status: GAGAL")
```
